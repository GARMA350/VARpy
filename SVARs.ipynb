```python
import pandas as pd
import numpy as np
from scipy import stats

# =========================================================
# FUNCIÓN OLS CON ESTADÍSTICOS
# =========================================================
def R_MCO(VY, MVX, nivel_confianza=95):

    y = VY.iloc[:, 0].to_numpy(dtype=float)

    X = np.column_stack([
        np.ones(len(y)),
        MVX.to_numpy(dtype=float)
    ])

    nombres = ['Constante'] + list(MVX.columns)

    n = len(y)
    k = X.shape[1] - 1
    gl = n - k - 1

    XtX = X.T @ X
    XtX_inv = np.linalg.inv(XtX)
    beta = XtX_inv @ (X.T @ y)

    y_hat = X @ beta
    resid = y - y_hat

    SCE = resid.T @ resid
    SCR = np.sum((y_hat - y.mean())**2)
    SCT = np.sum((y - y.mean())**2)

    R2 = 1 - SCE / SCT
    R2_aj = 1 - (SCE / gl) / (SCT / (n - 1))

    sigma2 = SCE / gl
    sigma = np.sqrt(sigma2)

    VCOV = sigma2 * XtX_inv
    se = np.sqrt(np.diag(VCOV))

    t_stats = beta / se
    p_values = 2 * (1 - stats.t.cdf(np.abs(t_stats), gl))

    alpha = 1 - nivel_confianza / 100
    t_crit = stats.t.ppf(1 - alpha / 2, gl)

    F = (SCR / k) / (SCE / gl)
    p_F = 1 - stats.f.cdf(F, k, gl)

    resultados = pd.DataFrame({
        'Coeficiente': beta,
        'Error estándar': se,
        't': t_stats,
        'P>|t|': p_values
    }, index=nombres)

    resumen = pd.Series({
        'Observaciones': n,
        'Regresores': k,
        'gl residuos': gl,
        'R2': R2,
        'R2 ajustado': R2_aj,
        'Sigma': sigma,
        'F': F,
        'Prob(F)': p_F
    })

    return resultados, resumen, beta, VCOV, resid, y_hat


# =========================================================
# 1. CARGAR DATOS
# =========================================================
df = pd.read_excel('/content/Primer_Data_ad.xlsx')

df['trimestre'] = pd.PeriodIndex(df['Unnamed: 0'], freq='Q')
df = df.set_index('trimestre').drop(columns=['Unnamed: 0'])

df = df.apply(pd.to_numeric, errors='coerce')

# =========================================================
# 2. CREAR REZAGOS
# =========================================================
def crear_rezagos(data, p):

    rezagos = []

    for j in range(1, p + 1):
        temp = data.shift(j)
        temp.columns = [f'{c}_lag{j}' for c in data.columns]
        rezagos.append(temp)

    return pd.concat(rezagos, axis=1)

p = 2  # número de rezagos

Y = df.copy()

X_lags = crear_rezagos(Y, p)

# Combinar y eliminar NaN
data_var = pd.concat([Y, X_lags], axis=1).dropna()

Y_clean = data_var[Y.columns]
X_clean = data_var[X_lags.columns]

# =========================================================
# 3. ESTIMAR VAR ECUACIÓN POR ECUACIÓN
# =========================================================
resultados_VAR = {}
residuos_VAR = pd.DataFrame(index=Y_clean.index)

print('='*80)
print('ESTIMACIÓN DEL VAR REDUCIDO')
print('='*80)

for variable in Y_clean.columns:

    print(f'\n\nECUACIÓN DEPENDIENTE: {variable}')
    print('-'*80)

    VY = Y_clean[[variable]]

    resultados, resumen, beta, VCOV, resid, y_hat = R_MCO(
        VY,
        X_clean,
        nivel_confianza=95
    )

    # Guardar resultados
    resultados_VAR[variable] = {
        'tabla': resultados,
        'resumen': resumen,
        'beta': beta,
        'VCOV': VCOV
    }

    residuos_VAR[variable] = resid

    # Mostrar resultados
    display(resultados)

# =========================================================
# 4. MATRICES DINÁMICAS A_j
# =========================================================
k = len(Y_clean.columns)

A_matrices = {}

for j in range(1, p + 1):

    A_j = np.zeros((k, k))

    for i, dep in enumerate(Y_clean.columns):

        beta = resultados_VAR[dep]['beta']

        # quitar constante
        coef_lags = beta[1:]

        start = (j - 1) * k
        end = j * k

        A_j[i, :] = coef_lags[start:end]

    A_j = pd.DataFrame(
        A_j,
        index=Y_clean.columns,
        columns=Y_clean.columns
    )

    A_matrices[j] = A_j

# Mostrar matrices A_j
print('\n' + '='*80)
print('MATRICES DINÁMICAS')
print('='*80)

for j, A_j in A_matrices.items():
    print(f'\nA{j}:')
    display(A_j)

# =========================================================
# 5. MATRIZ SIGMA_u
# =========================================================
T = residuos_VAR.shape[0]
m = X_clean.shape[1] + 1  # constante

Sigma_u = (residuos_VAR.T @ residuos_VAR) / (T - m)

Sigma_u = pd.DataFrame(
    Sigma_u,
    index=Y_clean.columns,
    columns=Y_clean.columns
)

print('\n' + '='*80)
print('MATRIZ DE COVARIANZAS DE LOS ERRORES (Sigma_u)')
print('='*80)

display(Sigma_u)

# =========================================================
# 6. MATRIZ COMPAÑERA GENERAL
# =========================================================

def construir_matriz_companera(A_matrices):

    p = len(A_matrices)

    # Convertir A_j a ndarray
    A_list = [np.asarray(A_matrices[j]) for j in range(1, p + 1)]

    # Número de variables
    k = A_list[0].shape[0]

    # -----------------------------------------------------
    # Primera fila por bloques: [A1 A2 ... Ap]
    # -----------------------------------------------------
    top = np.hstack(A_list)

    # -----------------------------------------------------
    # Caso VAR(1)
    # -----------------------------------------------------
    if p == 1:
        Phi_comp = top

    # -----------------------------------------------------
    # Caso general VAR(p)
    # -----------------------------------------------------
    else:

        # Identidad desplazada
        bottom_left = np.eye(k * (p - 1))

        # Bloque de ceros final
        bottom_right = np.zeros((k * (p - 1), k))

        bottom = np.hstack([bottom_left, bottom_right])

        Phi_comp = np.vstack([top, bottom])

    return Phi_comp


# Construir matriz compañera
Phi_comp = construir_matriz_companera(A_matrices)


# =========================================================
# MOSTRAR RESULTADO
# =========================================================

print('\n' + '='*80)
print('MATRIZ COMPAÑERA')
print('='*80)

Phi_comp_df = pd.DataFrame(Phi_comp)

display(Phi_comp_df)

print(f'Dimensión: {Phi_comp.shape}')
```

```python
import numpy as np

# =========================================================
# 1. MATRIZ INICIAL DE COVARIANZAS
# =========================================================

P = np.linalg.cholesky(Sigma_u.to_numpy())

# Número de variables
k = len(Y_clean.columns)

# Número de rezagos (ya definido antes)
# p = 2


# =========================================================
# 2. MATRIZ ORTOGONAL ALEATORIA
# =========================================================

def random_orthogonal(k):

    Z = np.random.randn(k, k)

    Q, R = np.linalg.qr(Z)

    # Convención de signos
    D = np.diag(np.sign(np.diag(R)))

    Q = Q @ D

    return Q


# =========================================================
# 3. FUNCIÓN IRF CON MATRIZ COMPAÑERA
# =========================================================

def irf_var_companion(Phi_comp, B, k, p, H=20):

    # Construir matriz de impacto compañera
    if p == 1:
        B_comp = B
    else:
        B_comp = np.vstack([
            B,
            np.zeros((k*(p-1), k))
        ])

    IR = np.zeros((H+1, k, k))

    Phi_h = np.eye(k*p)

    for h in range(H+1):

        response = Phi_h @ B_comp

        # Tomar sólo las variables originales
        IR[h] = response[:k, :]

        Phi_h = Phi_h @ Phi_comp

    return IR


# =========================================================
# 4. RESTRICCIONES DE SIGNO SOBRE B (IMPACTO)
# =========================================================

# Modificación: La función ahora acepta una matriz de restricciones R
def sign_ok_B(B, R):
    """
    Verifica las restricciones de signo sobre la matriz de impacto B
    basado en la matriz de restricciones R.
    R[i, j] = 1  => B[i, j] > 0
    R[i, j] = -1 => B[i, j] < 0
    R[i, j] = 0  => No hay restricción
    """
    k_rows, k_cols = B.shape
    if R.shape != (k_rows, k_cols):
        raise ValueError("Dimensiones de la matriz R deben coincidir con las de B")

    for i in range(k_rows):
        for j in range(k_cols):
            if R[i, j] == 1 and B[i, j] <= 0:
                return False
            if R[i, j] == -1 and B[i, j] >= 0:
                return False
    return True

# Ejemplo de matriz de restricciones R, que replica las restricciones originales:
# choque monetario = columna 1 (1-year T-Bill shock)
# PIB (fila 0) negativo ante choque monetario
# tasa (fila 1) positiva ante choque monetario
# Adicionalmente, se replicarán las otras restricciones implícitas en el código original:
# PIB (fila 0) ante choque PIB (columna 0) positivo
# Tasa (fila 1) ante choque PIB (columna 0) positivo
R_restrictions = np.array([
    [1, -1],  # B[0,0] > 0 (PIB vs PIB shock), B[0,1] < 0 (PIB vs T-Bill shock)
    [1,  1]    # B[1,0] > 0 (T-Bill vs PIB shock), B[1,1] > 0 (T-Bill vs T-Bill shock)
])


# =========================================================
# 5. ALGORITMO DE ACEPTACIÓN (SÓLO B)
# =========================================================

accepted_B = []

ndraws = 1000

for _ in range(ndraws):

    # Rotación ortogonal
    Q = random_orthogonal(k)

    # Matriz estructural candidata
    B_sr = P @ Q

    # Verificar restricciones sobre B usando la nueva función
    if sign_ok_B(B_sr, R_restrictions):

        accepted_B.append(B_sr)

accepted_B = np.array(accepted_B)

print('Rotaciones aceptadas:', len(accepted_B))


# =========================================================
# 6. CALCULAR IRF DE LAS ROTACIONES ACEPTADAS
# =========================================================

H = 50

accepted_IR = []

for B_sr in accepted_B:

    IR = irf_var_companion(
        Phi_comp=Phi_comp,
        B=B_sr,
        k=k,
        p=p,
        H=H
    )

    accepted_IR.append(IR)

accepted_IR = np.array(accepted_IR)

print('Dimensión accepted_IR:', accepted_IR.shape)


# =========================================================
# 7. MEDIANA PUNTUAL
# =========================================================

IR_med = np.median(accepted_IR, axis=0)

print('Dimensión IR_med:', IR_med.shape)


# =========================================================
# 8. FRY–PAGAN
# =========================================================

dist = np.sum(
    (accepted_IR - IR_med)**2,
    axis=(1,2,3)
)

idx_fp = np.argmin(dist)

B_fp = accepted_B[idx_fp]
IR_fp = accepted_IR[idx_fp]

print('Índice Fry–Pagan:', idx_fp)


# =========================================================
# 9. VERIFICACIÓN DE LAS RESTRICCIONES SOBRE B_fp
# =========================================================

print('\nB_fp:')
print(np.round(B_fp, 4))

print('\nRestricciones (verificadas con la matriz R):')
# Re-verificar con la nueva función y la matriz R_restrictions
print('Restricciones de R_restrictions cumplidas:', sign_ok_B(B_fp, R_restrictions))
```
